```python
storage_account_name = "bankingprojectstacc"
storage_account_key = dbutils.secrets.get(scope = 'keyvault-scope', key = 'storage-access-key')

container_name = "gold"
container_names= "silver"

spark.conf.set(f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net", storage_account_key )
```

```python
fct_transaction = spark.table("db_bankingproject.db_bankingproject.fct_banking_transactions")

display(fct_transaction)
```

```python

folder_name = "fct_banking_transactions"
output_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{folder_name}"
fct_transaction.write \
    .mode("overwrite") \
        .format("parquet") \
            .save(output_path)
```
